### Importando dados de um arquivo XLSX

Para importar dados de um arquivo Excel (`.xlsx`), vamos usar a biblioteca `pandas`. Primeiro, precisamos instalá-la se ainda não estiver instalada. Em seguida, usaremos a função `read_excel` para carregar o arquivo em um DataFrame.

Agora, vamos exibir as primeiras 5 linhas do DataFrame para ter uma ideia dos dados.

```python
if 'df' in locals(): # Verifica se o DataFrame foi criado com sucesso
    display(df.head())
else:
    print("Não foi possível exibir o DataFrame porque o arquivo não foi carregado.")
```

```python
import pandas as pd
df = pd.read_excel('dados.xlsx')
display(df.head(6))
display(df.tail(6))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get the value counts for the specified column
fruit_preference_counts = df['Qual dessas frutas você prefere para suco natural?'].value_counts().reset_index()
fruit_preference_counts.columns = ['Fruta', 'Contagem']

# Create the bar plot
plt.figure(figsize=(5, 5))
sns.barplot(x='Fruta', y='Contagem', data=fruit_preference_counts, palette='viridis')
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Respostas')
plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```
